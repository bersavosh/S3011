# Simple Holoviews examples

Holoviews interface is very different than Seaborn or matplotlib.

However, it enables rapid interactive visualizations prepared for deployment within browsers.

```python
import numpy as np
import pandas as pd
import holoviews as hv
from holoviews import opts, dim
from bokeh.models import HoverTool
hv.extension('bokeh')

DATA_DIR = 'https://raw.githubusercontent.com/bersavosh/S3011/main/data/'
```

Continuing with the loan data we have worked on in the pandas and seaborn examples:

```python
data = pd.read_csv(DATA_DIR+'loan_data_subset.csv')
data.columns
```

Holoviews has its own data interface, which when we declare our data within it, allows rapid declarative visualization with faceting and built-in interactions already activated.

```python
ds = hv.Dataset(data,
                kdims=['agency_abbr','loan_purpose_name','applicant_sex_name'],
                vdims=['applicant_income_000s','loan_amount_000s'],)

display(ds)
```

##  a bar chart categorizing with both agency and loan type

(We made this in seaborn, let's make it here again for comparison)

```python
bars = ds.to.bars(['loan_purpose_name','agency_abbr'],'loan_amount_000s',groupby=[]).aggregate(function=len)
bars.opts(opts.Bars(width=800, height=400, tools=['hover'], xrotation=90, show_legend=False))
```

## A simple dashboard in Holoviews

```python
scatter = ds.to.points(['applicant_income_000s','loan_amount_000s'],vdims=['loan_purpose_name'],groupby='applicant_sex_name')
scatter.opts(opts.Points(cmap='Dark2',size=3,color=dim('loan_purpose_name'),logx=True,logy=True,width=500, height=400, tools=['hover']))
```

```python
scatter = ds.to.points(['applicant_income_000s','loan_amount_000s'],vdims=['loan_purpose_name','applicant_sex_name'],groupby=[])
scatter.opts(opts.Points(cmap='Dark2',size=3,color=dim('loan_purpose_name'),
                         logx=True,logy=True,width=500, height=400, tools=['hover']))
```
